```python
import json
import pandas as pd

# Load dataset
file_path = "aspect-dataset.json"  # update path if needed
with open(file_path, "r", encoding="utf-8") as f:
    data = json.load(f)

# Convert to DataFrame
df = pd.DataFrame(data)

# Total entries
print("Total entries:", len(df))

# Unique aspect values and their counts
aspect_counts = df['aspect'].value_counts()

print("\nUnique aspects and their counts:")
print(aspect_counts)
```
